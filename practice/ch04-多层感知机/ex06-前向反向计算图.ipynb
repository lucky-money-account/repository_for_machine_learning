## 题 1 🌱 手算前向传播

给定一个小网络（2 输入 → 2 隐藏 → 1 输出，ReLU 激活）：

- x = [1, 2]
- W1 = [[1, 2], [3, 4]]，b1 = [0.5, 0.5]
- W2 = [[1], [2]]，b2 = [0]
- 目标 y = 30，损失用 MSE 的一半：(ŷ−y)²/2

先手算：隐藏层 H、输出 out、损失 loss（写出完整过程），再运行对照。

**【你的预测】**

```python
import torch

x = torch.tensor([1.0, 2.0])
W1 = torch.tensor([[1.0, 2.0], [3.0, 4.0]], requires_grad=True)
b1 = torch.tensor([0.5, 0.5], requires_grad=True)
W2 = torch.tensor([[1.0], [2.0]], requires_grad=True)
b2 = torch.tensor([0.0], requires_grad=True)
y = torch.tensor([30.0])

H = torch.relu(x @ W1 + b1)     # 手算?
out = H @ W2 + b2               # 手算?
l = ((out - y) ** 2) / 2        # 手算?
print('H   =', H.tolist())
print('out =', out.item())
print('loss=', l.item())
```

## 题 2 🔧 手算反向传播（链式法则）

先手算每个参数的梯度（写出链式法则的每一步），再运行对照：

- dL/dW2、dL/db2
- dL/dW1、dL/db1（注意 ReLU 的导数）

提示：dL/dout = (out − y)；dL/dW2 = Hᵀ · dL/dout；dL/dH = W2ᵀ · dL/dout；再乘 ReLU 的导数。

**【你的预测】**

```python
l.backward()
print('dL/dW2 =', W2.grad.tolist())
print('dL/db2 =', b2.grad.tolist())
print('dL/dW1 =', W1.grad.tolist())
print('dL/db1 =', b1.grad.tolist())
```
