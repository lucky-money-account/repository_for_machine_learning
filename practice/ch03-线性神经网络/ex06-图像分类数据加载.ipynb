```python
%matplotlib inline
import torch
import torchvision
from torch.utils import data
from torchvision import transforms
import matplotlib.pyplot as plt

trans = transforms.ToTensor()
mnist_train = torchvision.datasets.FashionMNIST(
    root='../data', train=True, transform=trans, download=True)
mnist_test = torchvision.datasets.FashionMNIST(
    root='../data', train=False, transform=trans, download=True)
print('训练集样本数:', len(mnist_train))
print('测试集样本数:', len(mnist_test))
```

```python
class_names = ['t-shirt', 'trouser', 'pullover', 'dress', 'coat',
               'sandal', 'shirt', 'sneaker', 'bag', 'ankle boot']
idx = 0
print('去掉 batch 维和通道维后的形状:', list(X[idx, 0].shape))
plt.imshow(X[idx, 0].numpy(), cmap='gray')
plt.title(f'label {y[idx].item()} = {class_names[y[idx].item()]}')
plt.axis('off')
plt.show()
```

```python
# 显示前 10 个样本（2×5 网格）
fig, axes = plt.subplots(2, 5, figsize=(10, 4))
for ax, i in zip(axes.flatten(), range(10)):
    ax.imshow(X[i, 0].numpy(), cmap='gray')
    ax.set_title(class_names[y[i].item()], fontsize=9)
    ax.axis('off')
plt.tight_layout()
plt.show()
```

## 小结与面试衔接

- NCHW 约定：批量数、通道数、高、宽（面试高频，torchvision 默认排布）
- ToTensor：PIL 图 → float 张量并除以 255，数值落在 [0, 1]
- 训练集 shuffle 打乱顺序，防止模型记住样本顺序；测试集顺序不影响评估
- Dataset 管单样本访问（__getitem__），DataLoader 管批量、打乱、并行读取
