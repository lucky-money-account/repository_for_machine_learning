## 题 3 🌱 `rand` 与 `randn` 的分布差异

各采样 1000 个数，观察它们的**取值范围**。运行后回答：为什么 `rand` 的值都在 [0, 1) 内，而 `randn` 会有负数？

```python
a = torch.rand(1000)
b = torch.randn(1000)
print('rand  : min =', a.min().item(), ' max =', a.max().item(), ' mean =', a.mean().item())
print('randn : min =', b.min().item(), ' max =', b.max().item(), ' mean =', b.mean().item())
```

```python
help(torch.randn)
```

**【你的观察结论】**

    on the interval :math:`[0, 1)`）
torch.randn就没有这样的特征
（Returns a tensor filled with random numbers from a normal distribution
with mean `0` and variance `1` (also called the standard normal distribution).）

## 题 4 🌱 二维 `cat` 形状预测

`A`、`B` 都是 `(2, 3)`。预测 `dim=0` 和 `dim=1` 拼接后的形状与内容。

**【你的预测】**

- `dim=0` → 形状 （4，3），内容：2行+2行叠起来
- `dim=1` → 形状 （2，6），内容：3列+3列叠起来

```python
A = torch.tensor([[1., 2., 3.], [4., 5., 6.]])
B = torch.tensor([[7., 8., 9.], [10., 11., 12.]])
print(torch.cat((A, B), dim=0), '\n')
print(torch.cat((A, B), dim=1))
```

## 题 5 🔧 三维 `cat` 形状预测

`A`、`B` 都是 `(2, 3, 4)`。预测三个维度各自拼接后的形状。

**【你的预测】**

- `dim=0` → (4, 3, 4)
- `dim=1` → (2, 6, 4)
- `dim=2` → (2, 3, 8)

```python
A = torch.ones(2, 3, 4)
B = torch.zeros(2, 3, 4)
print('dim=0:', torch.cat((A, B), dim=0).shape)
print('dim=1:', torch.cat((A, B), dim=1).shape)
print('dim=2:', torch.cat((A, B), dim=2).shape)
```

## 题 8 🔧 原地操作 vs 非原地操作

运行后回答：`Y = X + 2` 之后 `X` 变了吗？`X += 2` 之后呢？再试试 `X.add_(2)`（带下划线 `_`）。

**【你的预测】**

- `Y = X + 2` 后 X：不变
- `X += 2` 后 X：变

```python
X = torch.zeros(2, 3)
print('初始 X =', X)
Y = X + 2
print('Y = X + 2 之后，X =', X)   # X 变了没？
X += 2                             # 等价于 X.add_(2)，原地操作
print('X += 2 之后，X =', X)       # 这次呢？
X -= 2
X.add_(2)
print('X.add_(2) = ',X)
```

**【你的预测】**第 2 步取出的形状是 ____，第 3 步 permute 的参数是 ____

```python
help(torch.permute)
```

```python
# 1. 创建图像 batch，用 arange 填充以便区分像素
imgs = torch.arange(2 * 3 * 4 * 4).reshape(2, 3, 4, 4)
print('imgs.shape =', imgs.shape)

# 2. 取出第 1 张图的 R 通道（写出你的代码）
R = imgs[0,0,:,:]
print('img.1R = ', R)
# 3. 转成 HWC 格式 (2, 4, 4, 3)（写出你的代码）
imgs = torch.permute(imgs,(0,2,3,1))
print('HWC:', imgs)
```
