```python
# ==== 测试辅助函数（不用改，运行即可）====

def check(name, fn):
    """执行一个测试项，统一处理未完成 / 答错 / 异常三种情况。"""
    try:
        fn()
        print(f'✓ {name}')
    except NotImplementedError as e:
        print(f'⚠ {name} 还没做：{e}')
    except AssertionError as e:
        print(f'✗ {name} 结果不对：{e}')
    except Exception as e:
        print(f'✗ {name} 出错了：{type(e).__name__}: {e}')

def assert_eq(got, expected):
    assert got == expected, f'期望 {expected}，得到 {got}'

def assert_close(got, expected, tol=1e-6):
    assert all(abs(g - e) < tol for g, e in zip(got, expected)), f'期望 {expected}，得到 {got}'

def assert_close2d(got, expected, tol=1e-6):
    assert len(got) == len(expected), f'行数不一致: 期望 {len(expected)} 行，得到 {len(got)} 行'
    for g_row, e_row in zip(got, expected):
        assert len(g_row) == len(e_row), f'列数不一致: 期望 {len(e_row)}，得到 {len(g_row)}'
        assert all(abs(g - e) < tol for g, e in zip(g_row, e_row)), f'期望 {expected}，得到 {got}'

def expect_error(fn, err, why):
    """断言 fn 会抛出 err 类型的异常。"""
    try:
        fn()
    except err:
        return
    raise AssertionError(f'应该报 {err.__name__}（{why}），但没有报错')
```
